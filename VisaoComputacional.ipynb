# Sobre o Projeto

O reconhecimento automático de expressões faciais é uma das principais aplicações da Visão Computacional e da Inteligência Artificial, sendo utilizado em sistemas de monitoramento, interação humano-computador, robótica, saúde e análise comportamental.

Este projeto apresenta um pipeline completo capaz de realizar a detecção de pessoas, localizar automaticamente a região da face e classificar emoções em tempo real utilizando webcam ou vídeos.

Para isso, foram integradas duas abordagens:

- **YOLOv8 Pose**, responsável pela detecção das pessoas e dos pontos anatômicos do rosto.
- **ExpressionAnyNet**, uma Rede Neural Convolucional personalizada baseada na arquitetura AnyNet, responsável pela classificação das emoções.

Além do desenvolvimento da arquitetura, foram realizadas etapas de otimização automática dos hiperparâmetros utilizando **Optuna** e monitoramento completo do treinamento utilizando **Weights & Biases (W&B)**.

````python
# ==========================================
# FASE 1: PREPARAÇÃO DOS DATALOADERS
# ==========================================
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Definir as transformações para as imagens (112x112)
train_transforms = transforms.Compose([
    transforms.Resize((112, 112)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

val_test_transforms = transforms.Compose([
    transforms.Resize((112, 112)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Função que o Optuna vai usar para testar diferentes Batch Sizes depois
def get_dataloaders(batch_size):
    train_dataset = datasets.ImageFolder(root='./ferplus_dataset/train', transform=train_transforms)
    val_dataset = datasets.ImageFolder(root='./ferplus_dataset/validation', transform=val_test_transforms)

    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)

    return train_loader, val_loader

# Teste rápido
train_loader, val_loader = get_dataloaders(64)
from collections import Counter

contador = Counter(train_loader.dataset.targets)

print("\nDistribuição das classes no treino:\n")

for i, nome in enumerate(train_loader.dataset.classes):
    print(f"{nome}: {contador[i]} imagens")

# Exibir a quantidade de imagens para confirmar
print(f"Dataset carregado com sucesso!")
print(f"Total de classes mapeadas automaticamente pelo PyTorch: {len(train_loader.dataset.classes)}")
print(f"Nomes das classes detectadas: {train_loader.dataset.classes}")
````

````python
import torch
import torch.nn as nn
import torch.nn.functional as F

# bloco construtor que se repete
class AnyNetBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super(AnyNetBlock, self).__init__()
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(out_channels)

        self.shortcut = nn.Sequential()
        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(out_channels)
            )

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += self.shortcut(x)
        return F.relu(out)

# estágio que agrupa vários blocos
class AnyNetStage(nn.Module):
    def __init__(self, in_channels, out_channels, num_blocks, stride):
        super(AnyNetStage, self).__init__()
        blocks = [AnyNetBlock(in_channels, out_channels, stride)]
        for _ in range(1, num_blocks):
            blocks.append(AnyNetBlock(out_channels, out_channels, stride=1))
        self.stage = nn.Sequential(*blocks)


    def forward(self, x):
        return self.stage(x)

# rede completa(posso alterar)
class ExpressionAnyNet(nn.Module):
    def __init__(self, config, num_classes=8):
        super(ExpressionAnyNet, self).__init__()

        # Stem
        self.stem = nn.Sequential(
            nn.Conv2d(
                3,
                config['stem_channels'],
                kernel_size=3,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(config['stem_channels']),
            nn.ReLU(inplace=True)
        )

        # Corpo
        current_channels = config['stem_channels']
        stages = []

        for i in range(len(config['stage_depths'])):
            stages.append(
                AnyNetStage(
                    current_channels,
                    config['stage_channels'][i],
                    config['stage_depths'][i],
                    stride=2
                )
            )
            current_channels = config['stage_channels'][i]

        self.body = nn.Sequential(*stages)

        # Cabeça
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        self.dropout = nn.Dropout(p=0.30)

        self.fc = nn.Linear(current_channels, num_classes)

    def forward(self, x):

        x = self.stem(x)

        x = self.body(x)

        x = self.pool(x)

        x = torch.flatten(x, 1)

        x = self.dropout(x)

        x = self.fc(x)

        return x
````

````python
# ==========================================
# FASE 1: FUNÇÕES DE TREINO E VALIDAÇÃO
# ==========================================
def train_one_epoch(model, dataloader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in dataloader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc

def validate(model, dataloader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in dataloader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

    val_loss = running_loss / total
    val_acc = correct / total
    return val_loss, val_acc
````

````python
# ==========================================
# FASE 1: FUNÇÃO OBJETIVO (OPTUNA + WANDB)
# ==========================================
!pip install -q optuna wandb
import optuna
import wandb
def objective(trial):
    # Definição do espaço de busca dos hiperparâmetros pela AnyNet
    num_stages = trial.suggest_int('num_stages', 2, 4) # Escolhe entre 2 e 4 estágios

    config = {
        'stem_channels': trial.suggest_categorical('stem_channels', [16, 32, 64]),
        'stage_depths': [trial.suggest_int(f'depth_stage_{i}', 1, 3) for i in range(num_stages)],
        'stage_channels': []
    }

    # Garantir que os canais dos estágios sejam crescentes (Conceito AnyNet/RegNet)
    channels_options = [32, 64, 128, 256, 512]
    start_idx = trial.suggest_int('start_channel_idx', 0, len(channels_options) - num_stages)
    config['stage_channels'] = channels_options[start_idx : start_idx + num_stages]

    # Hiperparâmetros de otimização de treino
    lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
    batch_size = trial.suggest_categorical('batch_size', [32, 64])

    # Inicializar o log do experimento no Weights & Biases
    run = wandb.init(
        project="anynet-ferplus-optimization",
        name=f"trial_{trial.number}",
        config={**config, "lr": lr, "batch_size": batch_size},
        reinit=True
    )

    # Preparar estruturas
    train_loader, val_loader = get_dataloaders(batch_size)
    model = ExpressionAnyNet(config=config, num_classes=8).to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)

    best_val_acc = 0.0

    # 4. Loop de treino reduzido para a otimização (ex: 3 épocas por tentativa para poupar tempo)
    epochs = 3
    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = validate(model, val_loader, criterion, device)

        # Envia as métricas da época atual para a nuvem do WandB
        wandb.log({
            "epoch": epoch,
            "train_loss": train_loss,
            "train_acc": train_acc,
            "val_loss": val_loss,
            "val_acc": val_acc
        })

        if val_acc > best_val_acc:
            best_val_acc = val_acc

    # Finaliza o registro do WandB desta tentativa específica
    run.finish()

    return best_val_acc

# ==========================================
# EXECUÇÃO DA OTIMIZAÇÃO
# ==========================================
# Criar o estudo focado em maximizar a acurácia de validação
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=5) # Executa 5 combinações diferentes para testar

print("\n Otimização concluída!")
print("Melhores parâmetros encontrados:", study.best_params)
````

````python
# =====================================================================
# FASE 2: TREINAMENTO DEFINITIVO COM OS MELHORES PARÂMETROS
# =====================================================================

import os
import torch
import torch.nn as nn
import torch.optim as optim
import wandb
from collections import Counter

# =====================================================
# Configurações fixas baseadas no resultado do Optuna
# =====================================================
BEST_CONFIG = {
    "num_stages": 3,
    "stem_channels": 32,
    "depths": [3, 2, 3],
    "start_channel_idx": 1,
    "lr": 0.0003079509492659248,
    "batch_size": 64,
    "epochs": 25
}
print("-> Iniciando Treinamento Definitivo com os parâmetros do Optuna...")

# =====================================================
# Inicializar o WandB
# =====================================================

run = wandb.init(
    project="anynet-ferplus-optimization",
    name="anynet_final_model",
    config=BEST_CONFIG,
    reinit=True
)

# =====================================================
# Criar DataLoaders
# =====================================================

train_loader, val_loader = get_dataloaders(
    batch_size=BEST_CONFIG["batch_size"]
)

# =====================================================
# Verificar distribuição das classes
# =====================================================

targets = train_loader.dataset.targets
contador = Counter(targets)

print("\n==============================")
print("Distribuição das classes")
print("==============================")

for i, nome in enumerate(train_loader.dataset.classes):
    print(f"{nome}: {contador[i]} imagens")

# =====================================================
# Calcular pesos automaticamente
# =====================================================

num_classes = len(train_loader.dataset.classes)

pesos = []

for i in range(num_classes):
    peso = len(targets) / contador[i]
    pesos.append(peso)

class_weights = torch.tensor(
    pesos,
    dtype=torch.float
).to(device)

print("\nPesos utilizados:")

for nome, peso in zip(train_loader.dataset.classes, class_weights):
    print(f"{nome}: {peso:.2f}")

# =====================================================
# Configuração da arquitetura
# =====================================================

canais_por_estagio = [32, 64, 128]

config_modelo = {
    "stem_channels": BEST_CONFIG["stem_channels"],
    "stage_depths": BEST_CONFIG["depths"],
    "stage_channels": canais_por_estagio
}

print(f"\nConfiguração final enviada à rede: {config_modelo}")

# =====================================================
# Criar modelo
# =====================================================

model = ExpressionAnyNet(
    config=config_modelo,
    num_classes=8
).to(device)

# =====================================================
# Loss com pesos
# =====================================================
criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=0.1
)
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='max',
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

best_val_acc = 0.0
checkpoint_path = "modelo_augmentation_68_40.pth"

patience = 5
counter = 0

# =====================================================
# Loop de Treinamento
# =====================================================

for epoch in range(BEST_CONFIG["epochs"]):

    print(f"\n======== ÉPOCA {epoch+1}/{BEST_CONFIG['epochs']} ========")

    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    val_loss, val_acc = validate(
        model,
        val_loader,
        criterion,
        device
    )

    print(f"Treino    | Loss: {train_loss:.4f} | Acc: {train_acc*100:.2f}%")
    print(f"Validação | Loss: {val_loss:.4f} | Acc: {val_acc*100:.2f}%")

    wandb.log({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "learning_rate": optimizer.param_groups[0]["lr"]
    })

    # Atualiza o scheduler após registrar os dados
    scheduler.step(val_acc)

    if val_acc > best_val_acc:

        best_val_acc = val_acc
        counter = 0

        torch.save({
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_acc": val_acc,
            "config": config_modelo
        }, checkpoint_path)

        print(f"Novo melhor modelo salvo! ({val_acc*100:.2f}%)")

    else:

        counter += 1
        print(f"Sem melhoria por {counter} época(s).")

        if counter >= patience:
            print(f"\nEarly Stopping ativado na época {epoch+1}.")
            break

print("\n===================================")
print("TREINAMENTO FINALIZADO")
print(f"Melhor acurácia de validação: {best_val_acc*100:.2f}%")
print(f"Modelo salvo em: {checkpoint_path}")
print("===================================")

run.finish()
````

````python
# =====================================================================
# EXTRA 2: TESTAR O MODELO COM UMA IMAGEM QUALQUER + TOP-3
# =====================================================================

import torch
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt

# Ordem exata das classes usada pelo ImageFolder no treinamento
class_names = [
    'Raiva',
    'Desprezo',
    'Nojo',
    'Medo',
    'Felicidade',
    'Neutro',
    'Tristeza',
    'Surpresa'
]

# Caminho da imagem que será testada
CAMINHO_IMAGEM = "Jennie.jpg"


def testar_imagem_externa(caminho_img):

    print("-> Carregando o modelo e preparando a imagem...")

    try:
        # =============================================================
        # Carregar o melhor modelo salvo
        # =============================================================

        checkpoint = torch.load(
            "modelo_augmentation_68_40.pth",
            map_location=device
        )

        model_test = ExpressionAnyNet(
            config=checkpoint["config"],
            num_classes=8
        ).to(device)

        model_test.load_state_dict(
            checkpoint["model_state_dict"]
        )

        model_test.eval()

        # =============================================================
        # Transformações usadas na validação
        # =============================================================

        transform_pipe = transforms.Compose([
            transforms.Resize((112, 112)),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]
            )
        ])

        # =============================================================
        # Abrir e transformar a imagem
        # =============================================================

        img_pil = Image.open(caminho_img).convert("RGB")
        img_tensor = transform_pipe(img_pil)
        img_tensor = img_tensor.unsqueeze(0).to(device)

        # =============================================================
        # Fazer a inferência
        # =============================================================

        with torch.no_grad():

            outputs = model_test(img_tensor)

            probabilities = torch.softmax(
                outputs,
                dim=1
            )[0]

            top_probabilities, top_indices = torch.topk(
                probabilities,
                k=3
            )

        # Transformar para listas comuns do Python
        top_probabilities = top_probabilities.cpu().tolist()
        top_indices = top_indices.cpu().tolist()

        # Primeira colocada
        classe_prevista = class_names[top_indices[0]]
        porcentagem = top_probabilities[0] * 100

        # Definir o nível de confiança ANTES de mostrar o título
        if porcentagem < 40:
            status = "Baixa confiança"
        else:
            status = "Confiança aceitável"

        # =============================================================
        # Mostrar a imagem
        # =============================================================

        plt.figure(figsize=(7, 6))
        plt.imshow(img_pil)

        plt.title(
            f"Previsão: {classe_prevista} ({porcentagem:.1f}%)\n{status}",
            fontsize=14,
            color="darkblue",
            weight="bold"
        )

        plt.axis("off")
        plt.show()

        # =============================================================
        # Mostrar o Top-3
        # =============================================================

        print("\n==============================")
        print("TOP-3 PREVISÕES")
        print("==============================")

        for posicao, (indice, probabilidade) in enumerate(
            zip(top_indices, top_probabilities),
            start=1
        ):
            nome_classe = class_names[indice]
            percentual = probabilidade * 100

            print(
                f"{posicao}º - "
                f"{nome_classe}: "
                f"{percentual:.1f}%"
            )

        print("\n[Resultado]")
        print(
            f"A previsão principal foi {classe_prevista}, "
            f"com {porcentagem:.1f}% de confiança."
        )
        print(f"Nível de confiança: {status}")

    except FileNotFoundError:

        print(
            f"\n[ERRO] O arquivo '{caminho_img}' não foi encontrado."
        )

        print(
            "Faça o upload da imagem no menu lateral do Colab "
            "e confira o nome do arquivo."
        )

    except Exception as erro:

        print(f"\n[ERRO] Não foi possível realizar o teste: {erro}")


# Executar o teste
testar_imagem_externa(CAMINHO_IMAGEM)
````

````python

# ===================================================================================
# FASE FINAL OTIMIZADA:
# DETECÇÃO MULTI-FACE COM YOLOv8 POSE + CLASSIFICAÇÃO COM EXPRESSION ANYNET
# ===================================================================================

import base64
import cv2
import numpy as np
import torch
import torchvision.transforms as transforms

from PIL import Image
from IPython.display import display, Javascript, JSON
from google.colab.output import register_callback
from ultralytics import YOLO


# ===================================================================================
# CONFIGURAÇÃO DO DISPOSITIVO
# ===================================================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# O YOLO aceita 0 para a primeira GPU ou "cpu"
yolo_device = 0 if torch.cuda.is_available() else "cpu"

print(f"Executando inferência em: {device}")


# ===================================================================================
# CARREGAMENTO DOS MODELOS
# ===================================================================================

print("Carregando o YOLOv8 Pose...")

model_yolo = YOLO("yolov8n-pose.pt")


print("Carregando o modelo ExpressionAnyNet...")

checkpoint = torch.load(
    "best_expression_anynet.pth",
    map_location=device
)

model_anynet = ExpressionAnyNet(
    config=checkpoint["config"],
    num_classes=8
).to(device)

model_anynet.load_state_dict(
    checkpoint["model_state_dict"]
)

model_anynet.eval()

print("Modelos carregados com sucesso!")


# ===================================================================================
# ORDEM CORRETA DAS CLASSES
# ===================================================================================
#
# Esta ordem precisa ser exatamente igual à ordem utilizada pelo ImageFolder:
#
# ['angry', 'contempt', 'disgust', 'fear',
#  'happy', 'neutral', 'sad', 'suprise']
#
# ===================================================================================

class_names = [
    "Raiva",
    "Desprezo",
    "Nojo",
    "Medo",
    "Felicidade",
    "Neutro",
    "Tristeza",
    "Surpresa"
]


# ===================================================================================
# TRANSFORMAÇÕES UTILIZADAS NA IMAGEM DO ROSTO
# ===================================================================================

transform_pipe = transforms.Compose([

    # Garante uma imagem de três canais
    transforms.Grayscale(num_output_channels=3),

    # Mesmo tamanho utilizado no treinamento
    transforms.Resize((112, 112)),

    transforms.ToTensor(),

    # Normalização ImageNet
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ===================================================================================
# FUNÇÃO JAVASCRIPT RESPONSÁVEL PELA WEBCAM
# ===================================================================================

def iniciar_webcam_yolo():

    js = Javascript("""
    async function iniciarWebcamYOLO() {

        // ---------------------------------------------------------------
        // Elemento principal
        // ---------------------------------------------------------------

        const div = document.createElement('div');

        div.style.display = 'flex';
        div.style.flexDirection = 'column';
        div.style.alignItems = 'flex-start';
        div.style.gap = '10px';


        // ---------------------------------------------------------------
        // Botão para interromper a captura
        // ---------------------------------------------------------------

        const botao = document.createElement('button');

        botao.textContent = 'Interromper captura';

        botao.style.padding = '8px 15px';
        botao.style.fontSize = '14px';
        botao.style.cursor = 'pointer';


        // ---------------------------------------------------------------
        // Texto informativo
        // ---------------------------------------------------------------

        const status = document.createElement('div');

        status.textContent = 'Inicializando a câmera...';
        status.style.fontFamily = 'Arial';
        status.style.fontSize = '14px';


        // ---------------------------------------------------------------
        // Vídeo original escondido
        // ---------------------------------------------------------------

        const video = document.createElement('video');

        video.style.display = 'none';
        video.setAttribute('playsinline', '');


        // ---------------------------------------------------------------
        // Imagem que receberá o frame processado
        // ---------------------------------------------------------------

        const imgOutput = document.createElement('img');

        imgOutput.style.display = 'block';
        imgOutput.style.width = '640px';
        imgOutput.style.maxWidth = '100%';
        imgOutput.style.border = '2px solid #333';
        imgOutput.style.borderRadius = '6px';


        // ---------------------------------------------------------------
        // Adicionar elementos à tela
        // ---------------------------------------------------------------

        document.body.appendChild(div);

        div.appendChild(botao);
        div.appendChild(status);
        div.appendChild(imgOutput);
        div.appendChild(video);


        // ---------------------------------------------------------------
        // Solicitar acesso à webcam
        // ---------------------------------------------------------------

        let stream;

        try {

            stream = await navigator.mediaDevices.getUserMedia({

                video: {
                    width: {
                        ideal: 640
                    },

                    height: {
                        ideal: 480
                    },

                    facingMode: "user"
                },

                audio: false
            });

        } catch (erro) {

            status.textContent =
                'Não foi possível acessar a câmera: ' + erro.message;

            return;
        }


        video.srcObject = stream;

        await video.play();


        // ---------------------------------------------------------------
        // Canvas com resolução reduzida
        // ---------------------------------------------------------------

        const canvas = document.createElement('canvas');

        canvas.width = 640;
        canvas.height = 480;

        const contexto = canvas.getContext(
            '2d',
            {
                alpha: false
            }
        );


        // ---------------------------------------------------------------
        // Variável de controle
        // ---------------------------------------------------------------

        let capturaAtiva = true;

        status.textContent =
            'Câmera ativa — detectando pessoas e expressões faciais.';


        // ---------------------------------------------------------------
        // Encerrar a câmera
        // ---------------------------------------------------------------

        botao.onclick = () => {

            capturaAtiva = false;

            stream
                .getTracks()
                .forEach(track => track.stop());

            div.remove();
        };


        // ---------------------------------------------------------------
        // Ajustar a altura da área de saída do Colab
        // ---------------------------------------------------------------

        google.colab.output.setIframeHeight(
            document.documentElement.scrollHeight,
            true
        );


        // ---------------------------------------------------------------
        // Loop da webcam
        // ---------------------------------------------------------------

        while (capturaAtiva) {

            try {

                // Espelha a imagem horizontalmente, como uma webcam comum
                contexto.save();

                contexto.translate(
                    canvas.width,
                    0
                );

                contexto.scale(
                    -1,
                    1
                );

                contexto.drawImage(
                    video,
                    0,
                    0,
                    canvas.width,
                    canvas.height
                );

                contexto.restore();


                // JPEG com qualidade reduzida para diminuir a transferência
                const imgUri = canvas.toDataURL(
                    'image/jpeg',
                    0.60
                );


                // Enviar o frame ao Python
                const resposta =
                    await google.colab.kernel.invokeFunction(

                        'notebook.processar_frame_yolo',

                        [imgUri],

                        {}
                    );


                // Receber o resultado processado
                if (
                    resposta &&
                    resposta.data &&
                    resposta.data['application/json']
                ) {

                    const resultado =
                        resposta.data['application/json'];


                    if (resultado.img_desenhada) {

                        imgOutput.src =
                            resultado.img_desenhada;
                    }


                    if (resultado.erro) {

                        console.error(
                            resultado.erro
                        );

                        status.textContent =
                            'Erro: ' + resultado.erro;
                    }
                }


                // Pequena pausa para evitar sobrecarga
                // 120 ms deixa o sistema mais estável no Colab.
                await new Promise(
                    resolve => setTimeout(resolve, 120)
                );

            } catch (erro) {

                console.error(erro);

                status.textContent =
                    'Erro durante o processamento: ' +
                    erro.message;

                await new Promise(
                    resolve => setTimeout(resolve, 500)
                );
            }
        }
    }


    iniciarWebcamYOLO();
    """)

    display(js)


# ===================================================================================
# FUNÇÃO AUXILIAR PARA DESENHAR O RESULTADO
# ===================================================================================

def desenhar_resultado(
    frame,
    caixa,
    classe,
    porcentagem
):

    x1, y1, x2, y2 = caixa

    cor = (0, 255, 0)


    # Desenha o retângulo ao redor do rosto
    cv2.rectangle(
        frame,
        (x1, y1),
        (x2, y2),
        cor,
        2
    )


    texto = f"{classe} ({porcentagem:.1f}%)"


    fonte = cv2.FONT_HERSHEY_SIMPLEX

    escala_fonte = 0.55

    espessura = 1


    (largura_texto, altura_texto), baseline = (
        cv2.getTextSize(
            texto,
            fonte,
            escala_fonte,
            espessura
        )
    )


    # Evita desenhar o texto fora da imagem
    texto_x1 = max(0, x1)

    texto_y2 = max(
        altura_texto + 10,
        y1
    )

    texto_y1 = max(
        0,
        texto_y2 - altura_texto - 12
    )

    texto_x2 = min(
        frame.shape[1],
        texto_x1 + largura_texto + 10
    )


    # Fundo verde para o texto
    cv2.rectangle(
        frame,
        (texto_x1, texto_y1),
        (texto_x2, texto_y2),
        cor,
        -1
    )


    # Texto preto
    cv2.putText(
        frame,
        texto,
        (
            texto_x1 + 5,
            texto_y2 - 6
        ),
        fonte,
        escala_fonte,
        (0, 0, 0),
        espessura,
        cv2.LINE_AA
    )


# ===================================================================================
# CALLBACK PYTHON
# ===================================================================================

def processar_frame_yolo(img_base64):

    try:

        # ---------------------------------------------------------------------------
        # Converter o Base64 recebido em uma imagem OpenCV
        # ---------------------------------------------------------------------------

        _, encoded = img_base64.split(
            ",",
            1
        )

        dados = base64.b64decode(
            encoded
        )

        np_img = np.frombuffer(
            dados,
            dtype=np.uint8
        )

        frame = cv2.imdecode(
            np_img,
            cv2.IMREAD_COLOR
        )


        if frame is None:

            return JSON({
                "erro":
                    "Não foi possível decodificar o frame da webcam."
            })


        altura_frame, largura_frame = frame.shape[:2]


        # ---------------------------------------------------------------------------
        # Detectar pessoas e pontos anatômicos com YOLOv8 Pose
        # ---------------------------------------------------------------------------

        resultado_yolo = model_yolo.predict(

            source=frame,

            # Resolução menor para aumentar a velocidade
            imgsz=320,

            # Confiança mínima para detectar uma pessoa
            conf=0.40,

            # Evita muitas detecções sobrepostas
            iou=0.50,

            # Limite máximo de pessoas por frame
            max_det=10,

            # GPU ou CPU
            device=yolo_device,

            verbose=False

        )[0]


        # Listas para guardar os rostos encontrados
        rostos_tensor = []

        caixas_dos_rostos = []


        # ---------------------------------------------------------------------------
        # Extrair os pontos faciais
        # ---------------------------------------------------------------------------

        if (
            resultado_yolo.keypoints is not None and
            resultado_yolo.keypoints.xy is not None
        ):

            keypoints_data = (
                resultado_yolo
                .keypoints
                .xy
                .detach()
                .cpu()
                .numpy()
            )


            # Cada conjunto de keypoints representa uma pessoa
            for pontos_pessoa in keypoints_data:

                # YOLO Pose:
                #
                # 0 = nariz
                # 1 = olho esquerdo
                # 2 = olho direito
                # 3 = orelha esquerda
                # 4 = orelha direita

                pontos_rosto = pontos_pessoa[0:5]


                # Ignora pontos iguais a zero
                pontos_validos = pontos_rosto[
                    np.logical_and(
                        pontos_rosto[:, 0] > 0,
                        pontos_rosto[:, 1] > 0
                    )
                ]


                # Pelo menos dois pontos são necessários
                if len(pontos_validos) < 2:
                    continue


                # Limites dos pontos encontrados
                fx1, fy1 = np.min(
                    pontos_validos,
                    axis=0
                )

                fx2, fy2 = np.max(
                    pontos_validos,
                    axis=0
                )


                largura_pontos = fx2 - fx1

                altura_pontos = fy2 - fy1


                # Tamanho mínimo para impedir caixas minúsculas
                tamanho_base = max(
                    largura_pontos,
                    altura_pontos,
                    40
                )


                # Centro aproximado da face
                centro_x = (
                    fx1 + fx2
                ) / 2

                centro_y = (
                    fy1 + fy2
                ) / 2


                # Caixa ampliada para incluir testa, bochechas e mandíbula
                x1 = int(
                    max(
                        0,
                        centro_x - tamanho_base * 0.85
                    )
                )

                y1 = int(
                    max(
                        0,
                        centro_y - tamanho_base * 1.00
                    )
                )

                x2 = int(
                    min(
                        largura_frame,
                        centro_x + tamanho_base * 0.85
                    )
                )

                y2 = int(
                    min(
                        altura_frame,
                        centro_y + tamanho_base * 0.95
                    )
                )


                # Ignora caixas inválidas
                if x2 <= x1 or y2 <= y1:
                    continue


                rosto_crop = frame[
                    y1:y2,
                    x1:x2
                ]


                if rosto_crop.size == 0:
                    continue


                # Ignora rostos extremamente pequenos
                if (
                    rosto_crop.shape[0] < 20 or
                    rosto_crop.shape[1] < 20
                ):
                    continue


                # OpenCV utiliza BGR; PIL utiliza RGB
                rosto_rgb = cv2.cvtColor(
                    rosto_crop,
                    cv2.COLOR_BGR2RGB
                )


                rosto_pil = Image.fromarray(
                    rosto_rgb
                )


                rosto_tensor = transform_pipe(
                    rosto_pil
                )


                rostos_tensor.append(
                    rosto_tensor
                )

                caixas_dos_rostos.append(
                    (x1, y1, x2, y2)
                )


        # ---------------------------------------------------------------------------
        # Classificar todos os rostos encontrados em uma única operação
        # ---------------------------------------------------------------------------

        if len(rostos_tensor) > 0:

            lote_rostos = torch.stack(
                rostos_tensor
            ).to(
                device,
                non_blocking=True
            )


            with torch.inference_mode():

                saidas = model_anynet(
                    lote_rostos
                )

                probabilidades = torch.softmax(
                    saidas,
                    dim=1
                )

                confiancas, predicoes = torch.max(
                    probabilidades,
                    dim=1
                )


            # Converter para CPU uma única vez
            confiancas = (
                confiancas
                .detach()
                .cpu()
                .numpy()
            )

            predicoes = (
                predicoes
                .detach()
                .cpu()
                .numpy()
            )


            # Desenhar cada resultado
            for caixa, indice, confianca in zip(

                caixas_dos_rostos,
                predicoes,
                confiancas

            ):

                classe_prevista = class_names[
                    int(indice)
                ]

                porcentagem = float(
                    confianca * 100
                )


                desenhar_resultado(
                    frame=frame,
                    caixa=caixa,
                    classe=classe_prevista,
                    porcentagem=porcentagem
                )


        # ---------------------------------------------------------------------------
        # Codificar o frame processado em JPEG
        # ---------------------------------------------------------------------------

        sucesso, buffer = cv2.imencode(

            ".jpg",

            frame,

            [
                cv2.IMWRITE_JPEG_QUALITY,
                65
            ]
        )


        if not sucesso:

            return JSON({
                "erro":
                    "Não foi possível codificar o frame processado."
            })


        imagem_base64 = base64.b64encode(
            buffer
        ).decode(
            "utf-8"
        )


        img_uri_final = (
            "data:image/jpeg;base64,"
            + imagem_base64
        )


        return JSON({
            "img_desenhada":
                img_uri_final
        })


    except Exception as erro:

        print(
            "Erro no processamento:",
            erro
        )

        return JSON({
            "erro":
                str(erro)
        })


# ===================================================================================
# REGISTRAR O CALLBACK
# ===================================================================================

register_callback(
    "notebook.processar_frame_yolo",
    processar_frame_yolo
)


# ===================================================================================
# INICIAR A WEBCAM
# ===================================================================================

print("Iniciando webcam...")

iniciar_webcam_yolo()


````

````python
# ============================================================
# TESTE COM VÍDEO:
# YOLOv8 Pose + ExpressionAnyNet
# ============================================================

import cv2
import torch
import numpy as np
import torchvision.transforms as transforms

from PIL import Image
from IPython.display import Video, display


# ------------------------------------------------------------
# CONFIGURAÇÕES
# ------------------------------------------------------------

CAMINHO_VIDEO = "video.mp4"
VIDEO_SAIDA = "video_processado.mp4"

# Ordem das classes usada no treinamento
class_names = [
    "Raiva",
    "Desprezo",
    "Nojo",
    "Medo",
    "Felicidade",
    "Neutro",
    "Tristeza",
    "Surpresa"
]

# Transformação igual à validação/teste
transform_video = transforms.Compose([
    transforms.Resize((112, 112)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

model_anynet.eval()


# ------------------------------------------------------------
# FUNÇÃO PARA ENCONTRAR O ROSTO PELOS PONTOS DO YOLO POSE
# ------------------------------------------------------------

def obter_caixa_rosto(keypoints, largura, altura):
    """
    Usa os cinco primeiros pontos do YOLO Pose:

    0 = nariz
    1 = olho esquerdo
    2 = olho direito
    3 = orelha esquerda
    4 = orelha direita
    """

    pontos_rosto = keypoints[:5]

    # Remove pontos inválidos
    pontos_validos = []

    for x, y in pontos_rosto:
        if x > 0 and y > 0:
            pontos_validos.append((x, y))

    # Precisamos de pelo menos dois pontos válidos
    if len(pontos_validos) < 2:
        return None

    pontos_validos = np.array(pontos_validos)

    x_min = int(np.min(pontos_validos[:, 0]))
    x_max = int(np.max(pontos_validos[:, 0]))
    y_min = int(np.min(pontos_validos[:, 1]))
    y_max = int(np.max(pontos_validos[:, 1]))

    largura_rosto = x_max - x_min
    altura_rosto = y_max - y_min

    # Evita caixas muito pequenas
    if largura_rosto < 5 or altura_rosto < 5:
        return None

    # Aumenta a caixa ao redor dos pontos faciais
    margem_x = int(largura_rosto * 0.8)
    margem_superior = int(altura_rosto * 1.5)
    margem_inferior = int(altura_rosto * 2.5)

    x1 = max(0, x_min - margem_x)
    x2 = min(largura, x_max + margem_x)

    y1 = max(0, y_min - margem_superior)
    y2 = min(altura, y_max + margem_inferior)

    if x2 <= x1 or y2 <= y1:
        return None

    return x1, y1, x2, y2


# ------------------------------------------------------------
# FUNÇÃO PARA CLASSIFICAR UMA FACE
# ------------------------------------------------------------

def classificar_face(face_bgr):
    """
    Recebe o recorte do rosto em OpenCV/BGR
    e devolve a emoção e a confiança.
    """

    # OpenCV usa BGR, mas a PIL utiliza RGB
    face_rgb = cv2.cvtColor(face_bgr, cv2.COLOR_BGR2RGB)

    imagem_pil = Image.fromarray(face_rgb)

    entrada = transform_video(imagem_pil)
    entrada = entrada.unsqueeze(0).to(device)

    with torch.no_grad():
        saida = model_anynet(entrada)

        probabilidades = torch.softmax(saida, dim=1)

        confianca, indice = torch.max(probabilidades, dim=1)

    emocao = class_names[indice.item()]
    confianca = confianca.item()

    return emocao, confianca


# ------------------------------------------------------------
# ABRIR O VÍDEO
# ------------------------------------------------------------

video = cv2.VideoCapture(CAMINHO_VIDEO)

if not video.isOpened():
    raise FileNotFoundError(
        f"Não foi possível abrir o vídeo: {CAMINHO_VIDEO}"
    )

fps = video.get(cv2.CAP_PROP_FPS)

if fps <= 0:
    fps = 30

largura = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
altura = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))

print("Informações do vídeo:")
print(f"Resolução: {largura}x{altura}")
print(f"FPS: {fps:.2f}")
print(f"Quantidade de frames: {total_frames}")


# ------------------------------------------------------------
# PREPARAR O VÍDEO DE SAÍDA
# ------------------------------------------------------------

codec = cv2.VideoWriter_fourcc(*"mp4v")

gravador = cv2.VideoWriter(
    VIDEO_SAIDA,
    codec,
    fps,
    (largura, altura)
)


# ------------------------------------------------------------
# PROCESSAR FRAME POR FRAME
# ------------------------------------------------------------

frame_atual = 0

print("\nIniciando processamento...")

while True:

    sucesso, frame = video.read()

    if not sucesso:
        break

    frame_atual += 1

    # YOLO Pose detecta as pessoas e os pontos corporais
    resultados = model_yolo.predict(
        source=frame,
        conf=0.35,
        verbose=False
    )

    resultado = resultados[0]

    # Verifica se existem keypoints detectados
    if resultado.keypoints is not None:

        keypoints_detectados = resultado.keypoints.xy.cpu().numpy()

        for keypoints_pessoa in keypoints_detectados:

            caixa = obter_caixa_rosto(
                keypoints_pessoa,
                largura,
                altura
            )

            if caixa is None:
                continue

            x1, y1, x2, y2 = caixa

            face = frame[y1:y2, x1:x2]

            if face.size == 0:
                continue

            try:
                emocao, confianca = classificar_face(face)

                texto = f"{emocao}: {confianca * 100:.1f}%"

                # Caixa ao redor do rosto
                cv2.rectangle(
                    frame,
                    (x1, y1),
                    (x2, y2),
                    (0, 255, 0),
                    2
                )

                # Fundo do texto
                (largura_texto, altura_texto), _ = cv2.getTextSize(
                    texto,
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.65,
                    2
                )

                topo_texto = max(0, y1 - altura_texto - 12)

                cv2.rectangle(
                    frame,
                    (x1, topo_texto),
                    (x1 + largura_texto + 10, y1),
                    (0, 255, 0),
                    -1
                )

                cv2.putText(
                    frame,
                    texto,
                    (x1 + 5, y1 - 6),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.65,
                    (0, 0, 0),
                    2
                )

            except Exception as erro:
                print(f"Erro ao classificar uma face: {erro}")

    gravador.write(frame)

    # Mostra o progresso a cada 30 frames
    if frame_atual % 30 == 0:
        porcentagem = (
            frame_atual / total_frames * 100
            if total_frames > 0
            else 0
        )

        print(
            f"Processando: {frame_atual}/{total_frames} "
            f"({porcentagem:.1f}%)"
        )


# ------------------------------------------------------------
# FINALIZAR
# ------------------------------------------------------------

video.release()
gravador.release()

print("\nProcessamento concluído!")
print(f"Vídeo salvo como: {VIDEO_SAIDA}")
````
